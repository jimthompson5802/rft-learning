# GRPO Completion Exploration

This notebook loads one Parquet file from `grpo-arithmetic-lora-demo/completions` into a pandas DataFrame and previews selected records.

## Select A Completion File

Edit the filename below to choose which Parquet file to inspect.

```python
parquet_filename = "completions_00050.parquet"
records_to_show = 8
```

```python
from pathlib import Path

from IPython.display import Markdown, display

import pandas as pd
```

```python
pd.set_option("display.max_colwidth", None)
```

```python
completions_dir = Path("grpo-arithmetic-lora-demo/completions")
parquet_path = completions_dir / parquet_filename

if not parquet_path.exists():
    available_files = sorted(path.name for path in completions_dir.glob("*.parquet"))
    raise FileNotFoundError(
        f"Parquet file not found: {parquet_path}. Available files: {available_files}"
    )

parquet_path
```

```python
df = pd.read_parquet(parquet_path)
preview_columns = ["step", "prompt", "completion", "format_reward", "correctness_reward", "advantage"]
preview_df = df.head(records_to_show)[preview_columns]

print(
    f"Requested {records_to_show} record(s); file contains {len(df)} total record(s); "
    f"showing {len(preview_df)} record(s)."
)

for record_number, row in enumerate(preview_df.itertuples(index=False), start=1):
    display(Markdown(f"## Record {record_number}"))
    print(f"step: {row.step}")
    print(f"format_reward: {row.format_reward}")
    print(f"correctness_reward: {row.correctness_reward}")
    print(f"advantage: {row.advantage}")
    print("prompt:")
    print(row.prompt)
    print()
    print("completion:")
    print(row.completion)
    print("\n" + "-" * 80 + "\n")
```
